```python
import pandas as pd

# Đọc dữ liệu đã làm sạch
df = pd.read_csv("../data/processed/houses_clean.csv")

# Kiểm tra kích thước
print("Kích thước dữ liệu:", df.shape)

# Xem 5 dòng đầu
display(df.head())
```

```python
# Hiển thị tên các cột
print("Các cột dữ liệu:")

for col in df.columns:
    print("-", col)
```

```python
# Tạo tổng số phòng
df["total_rooms"] = (
    df["bedroom_count"] +
    df["bathroom_count"]
)

# Kiểm tra
display(
    df[
        [
            "bedroom_count",
            "bathroom_count",
            "total_rooms"
        ]
    ].head(10)
)
```

```python
# Tính diện tích trung bình cho mỗi phòng
df["area_per_room"] = (
    df["area"] /
    df["total_rooms"].replace(0, 1)
)

# Kiểm tra
display(
    df[
        [
            "area",
            "total_rooms",
            "area_per_room"
        ]
    ].head(10)
)
```

```python
# Xử lý ngày đăng tin
# Chuyển published_at sang kiểu ngày giờ
df["published_at"] = pd.to_datetime(
    df["published_at"],
    errors="coerce"
)

# Lấy năm đăng tin
df["published_year"] = df["published_at"].dt.year

# Lấy tháng đăng tin
df["published_month"] = df["published_at"].dt.month

# Lấy thứ trong tuần
df["published_dayofweek"] = df["published_at"].dt.dayofweek

# Kiểm tra
display(
    df[
        [
            "published_at",
            "published_year",
            "published_month",
            "published_dayofweek"
        ]
    ].head(10)
)
```

```python
# tạo độ dài mô tả 
# Tính số ký tự trong mô tả
df["description_length"] = (
    df["description"].str.len()
)

# Kiểm tra
display(
    df[
        [
            "description",
            "description_length"
        ]
    ].head()
)
```

```python
# tạo độ dài tên tin
# Tính số ký tự trong tên tin
df["name_length"] = df["name"].str.len()

# Kiểm tra
display(
    df[
        [
            "name",
            "name_length"
        ]
    ].head()
)
```

```python
# Danh sách đặc trưng mới
new_features = [
    "total_rooms",
    "area_per_room",
    "published_year",
    "published_month",
    "published_dayofweek",
    "description_length",
    "name_length"
]

# Hiển thị thống kê
print("CÁC ĐẶC TRƯNG MỚI:")

display(
    df[new_features].describe()
)
```

```python
# Kiểm tra kích thước
print("Kích thước sau Feature Engineering:", df.shape)

# Kiểm tra dữ liệu thiếu
print("\nDữ liệu thiếu:")

missing = df.isnull().sum()

print(
    missing[missing > 0]
)
```
