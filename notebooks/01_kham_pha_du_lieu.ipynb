```python
print("THỐNG KÊ GIÁ:")
print(df["price"].describe())
print("\nTHỐNG KÊ DIỆN TÍCH:")
print(df["area"].describe())
```

```python
print("=== KIỂM TRA GIÁ ===")

print("\nCác mức giá thấp nhất:")
print(df["price"].sort_values().head(20).to_string(index=False))

print("\nCác mức giá cao nhất:")
print(df["price"].sort_values(ascending=False).head(20).to_string(index=False))

print("\nCác phân vị của giá:")
print(df["price"].quantile([0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99, 0.995, 1.0]))
```

```python
print("=== KIỂM TRA OUTLIER GIÁ ===")

for nguong in [20e9, 30e9, 50e9, 100e9, 200e9, 500e9, 1e12]:
    so_luong = (df["price"] > nguong).sum()
    print(f"Giá > {nguong/1e9:.0f} tỷ: {so_luong} tin")
```

```python
print("=== NHỮNG TIN CÓ GIÁ CAO NHẤT ===")

cols = [
    "price",
    "area",
    "district_name",
    "ward_name",
    "property_type_name",
    "description"
]

cols_co = [c for c in cols if c in df.columns]

df.sort_values("price", ascending=False)[cols_co].head(20)
```

```python
print("=== 9 TIN CÓ GIÁ TRÊN 500 TỶ ===")

df_cao = df[df["price"] > 500e9].copy()

cols = [
    "price",
    "area",
    "district_name",
    "ward_name",
    "property_type_name",
    "bedroom_count",
    "bathroom_count",
    "floor_count",
    "description"
]

cols_co = [c for c in cols if c in df.columns]

print("Số tin:", len(df_cao))
display(
    df_cao[cols_co]
    .sort_values("price", ascending=False)
)
```
